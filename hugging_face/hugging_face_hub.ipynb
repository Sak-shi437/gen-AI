**Install the required Library**

```python
from transformers import pipeline
```

**Text Generation Pipeline**

```python
generator = pipeline(
    task = "text-generation",
    model = "Qwen/Qwen2.5-7B-Instruct",
    device_map = "auto"

)
```

Sample prompt

```python
prompt ="Explain AI"

result = generator(prompt, max_new_tokens=80, temperature=0.7)
print(result[0]['generated_text'])
```

```python
prompts = [
    'explain text classifier task in simple language',
    'explain text generation task in simple language',
    'explain text summerazition in simple language'
]
for prompt in prompts:
  result = generator(prompt, max_new_tokens=80, temperature=0.7)
  print(prompt)
  print(result[0]['generated_text'])
  print("-"*60)
```
